```python
import serial

PORT = "COM6"          # ganti kalau port ESP32 berbeda
arduinoData = serial.Serial(PORT, 115200, timeout=0)

try:
    while True:
        myData = input("Kirimkan perintah (1/0, q = keluar): ")
        if myData == "1":
            arduinoData.write(b"1")
        elif myData == "0":
            arduinoData.write(b"0")
        elif myData.lower() == "q":
            arduinoData.write(b"0")   # matikan motor sebelum keluar
            break
finally:
    arduinoData.close()
    print("Port ditutup.")
```
